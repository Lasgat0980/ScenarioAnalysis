```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
from IPython.display import display, Markdown

todays_date = datetime.now().strftime("%Y-%m-%d")
student_id = "Lasgat0980"

display(Markdown("### Step 2: Define Scenarios"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Define scenario parameters
principal = 300000
interest_rates = [0.03, 0.05, 0.08]
loan_terms = [15, 30, 40]

# Display scenario parameters
print("Principal Amount:", principal)
print("Interest Rates:", interest_rates)
print("Loan Terms:", loan_terms)
```

```python
display(Markdown("### Step 3: Perform the Scenario Analysis"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Define the monthly loan repayment function
def calculate_monthly_payment(principal, annual_rate, years):
    """Calculate the monthly loan repayment amount."""
    r = annual_rate / 12
    n = years * 12
    monthly_payment = principal * r * (1 + r)**n / ((1 + r)**n - 1)
    return monthly_payment

# Perform scenario analysis
results = []

for i in range(len(interest_rates)):
    monthly_payment = calculate_monthly_payment(principal, interest_rates[i], loan_terms[i])
    total_repayment = monthly_payment * loan_terms[i] * 12
    results.append({
        "Scenario": i + 1,
        "Interest_Rate": interest_rates[i],
        "Loan_Term": loan_terms[i],
        "Monthly_Payment": monthly_payment,
        "Total_Repayment": total_repayment
    })

# Convert results to DataFrame
results_df = pd.DataFrame(results)

# Display results
display(results_df)
```

```python
display(Markdown("### Step 4: Visualize the Results"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Plot total repayment amounts for each scenario
plt.figure(figsize=(12, 6))
plt.bar(results_df['Scenario'], results_df['Total_Repayment'])
plt.title("Total Loan Repayment by Scenario")
plt.xlabel("Scenario")
plt.ylabel("Total Repayment Amount ($)")
plt.xticks(results_df['Scenario'])
plt.show()
```

```python
display(Markdown("### Step 5: Apply Sensitivity Analysis"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Modify interest rates for sensitivity analysis
modified_interest_rates = [0.04, 0.06, 0.09]

# Perform scenario analysis with modified interest rates
modified_results = []

for i in range(len(modified_interest_rates)):
    monthly_payment = calculate_monthly_payment(principal, modified_interest_rates[i], loan_terms[i])
    total_repayment = monthly_payment * loan_terms[i] * 12
    modified_results.append({
        "Scenario": i + 1,
        "Interest_Rate": modified_interest_rates[i],
        "Loan_Term": loan_terms[i],
        "Monthly_Payment": monthly_payment,
        "Total_Repayment": total_repayment
    })

# Convert results to DataFrame
modified_results_df = pd.DataFrame(modified_results)

# Plot original and modified repayment amounts
plt.figure(figsize=(12, 6))
plt.bar(results_df['Scenario'] - 0.2, results_df['Total_Repayment'], width=0.4, label='Original Interest Rates')
plt.bar(modified_results_df['Scenario'] + 0.2, modified_results_df['Total_Repayment'], width=0.4, label='Modified Interest Rates')
plt.title("Loan Repayment Sensitivity Analysis")
plt.xlabel("Scenario")
plt.ylabel("Total Repayment Amount ($)")
plt.xticks(results_df['Scenario'])
plt.legend()
plt.show()
```

The scenarios showed that both the interest rate and the length of the loan affected the repayment amounts. Scenario 1 had a higher monthly payment than Scenario 2, but also had the lowest total repayment amount. Scenario 2 had the lowest monthly payment because the loan was spread out over 30 years. Scenario 3 had the highest total repayment amount because of the higher interest rate and longer loan term. Overall, the results showed that the lower monthly payments do not always mean paying less. 

When I increased the interest rates by 1% for each scenario, the total repayment amounts increased. The difference was noticeable in all three scenarios, but the longer loans showed a larger increase in the total amount paid. This tells us that even a small increase in interest rates can make a big difference in how much a loan costs over time.
